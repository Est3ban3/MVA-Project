```python
import ROOT, glob

pattern = "/Users/eviethorp/Desktop/Internship26/MVA-Project/trexfitter/results/*/Limits/Asymptotics/myLimit.root"

header = ["model", "-2sigma", "-1sigma", "median", "+1sigma", "+2sigma"]
print("{:<28}{:>10}{:>10}{:>10}{:>10}{:>10}".format(*header))

for fname in sorted(glob.glob(pattern)):
    label = fname.split("/results/")[1].split("/Limits/")[0]
    f = ROOT.TFile.Open(fname)
    t = f.Get("stats")
    t.GetEntry(0)
    print("{:<28}{:>10.2f}{:>10.2f}{:>10.2f}{:>10.2f}{:>10.2f}".format(
        label,
        t.exp_upperlimit_minus2, t.exp_upperlimit_minus1, t.exp_upperlimit,
        t.exp_upperlimit_plus1, t.exp_upperlimit_plus2))
    f.Close()
```
