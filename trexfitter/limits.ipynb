```python
import ROOT, glob

pattern = "/Users/eviethorp/Desktop/Internship26/MVA-Project/trexfitter/results/*/Limits/Asymptotics/myLimit.root"

for fname in sorted(glob.glob(pattern)):
    label = fname.split("/results/")[1].split("/Limits/")[0]
    f = ROOT.TFile.Open(fname)
    t = f.Get("stats")
    t.GetEntry(0)
    print(label, t.exp_upperlimit_minus2, t.exp_upperlimit_minus1, t.exp_upperlimit, t.exp_upperlimit_plus1, t.exp_upperlimit_plus2)
    f.Close()
    
#-2σ, -1σ, median, +1σ, +2σ
```
