# Export Model Scores to ROOT — TRExFitter inputs

Consolidates every model's out-of-fold (OOF) score CSV into **new** ROOT files,
**one file per (channel, run, process)**, each holding a flat `TTree` with one
branch per (model x track) score. This does **not** touch the original
per-process input ROOT files (`signal_ggF.root`, `ttbar.root`, ...) - it only
reads each model's own `oof_scores_*.csv` artifact and writes brand-new
`scores_<process>.root` files alongside them.

## Why this shape (three things that will silently break TRExFitter)

1. **`TTree`, not `RNTuple`.** The previous version wrote via
   `f[tree] = branches`, which in uproot >= 5.7 produces a `ROOT::RNTuple`.
   TRExFitter's ntuple reading is `TChain`/`TTree::Draw`-based and cannot open
   an RNTuple at all. We therefore use `mktree(...)` + `extend(...)`, which
   writes a real `TTree` (verified in the last cell).
2. **No string branches.** `process` used to be written as `std::string`.
   TRExFitter's `Selection` is a `TTree::Draw` expression and cannot cut on
   one; uproot's TTree writer refuses string dtypes outright
   (`cannot write NumPy dtype >U10 in TTree`). Splitting one file per process
   removes the need for the cut entirely - each `Sample` is just its own file,
   which is TRExFitter's native `NtupleFiles` idiom. A numeric `process_id` is
   carried anyway so the files stay self-describing if they are ever `hadd`ed.
3. **Positional alignment, never a key-based merge.** `eventNumber` is not a
   unique key (3,498 duplicated `(process, eventNumber)` rows measured in
   1l2tau/run2, multiplicity up to 3), so a join would fabricate rows. Every
   model's OOF file for a given (channel, run) is written from the SAME source
   DataFrame in the SAME row order, so scores are concatenated POSITIONALLY,
   guarded by asserts that fail loudly rather than corrupt the export.

## Branches written

| branch | meaning |
| --- | --- |
| `eventNumber`, `fold`, `label`, `process_id` | bookkeeping (`fold` is NOT a fit variable) |
| `w_phys` | full physical event weight - use as TRExFitter `MCweight` |
| `score_{xgboost,dnn,mlp,gnn}` | per-run-track models (trained on this run only) |
| `score_{xgboost,dnn,mlp,gnn}_comb` | combined-track models (trained on Run2+Run3 of this channel) |
| `score_pnn` | PNN (single network across both channels and both runs) |

The combined-track and PNN OOF files each cover several (channel, run)
combinations at once and carry their own `run` (and `channel`) columns, so they
are sliced rather than looked up by filename. **`run` is deliberately excluded
from the alignment key**: it is present in some per-run OOF csvs and absent in
others, while every combined-track csv has it.

A model whose OOF csv does not exist yet is simply **omitted** (never filled
with a sentinel) - referencing a missing branch then fails loudly in TRExFitter
instead of silently producing a garbage histogram. A warning lists them.

## Using these in TRExFitter

`w_phys` already includes cross-section x luminosity (the `weight` branch is a
per-campaign normalisation constant; `weights` is the per-event generator x
detector x efficiency correction). **Set `Lumi: 1.0`** - applying a luminosity
scale on top would inflate every yield by ~10^2.

```
Region: "SR_1l2tau_run2"
  Variable: "score_xgboost",15,0,1
Sample: "ttbar"
  NtupleFiles: "scores_ttbar"
  NtupleName: "nominal"
  MCweight: "w_phys"
```

Swap `Variable` to `score_dnn`, `score_pnn`, ... to get the limit per model
from one set of inputs. TRExFitter fits one discriminant at a time - the extra
branches are for cheap re-fits, not a combined multi-model fit.


````python
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import uproot

BASE = Path("PPSSP_2026")
CHANNELS = ["1l2tau", "2l2tau"]
RUNS = ["run2", "run3"]
TREE_NAME = "nominal"          # TRExFitter `NtupleName`

# Per-model OOF filename suffix. XGBoost is the canonical/unsuffixed artifact;
# every other model was suffixed specifically to avoid overwriting it.
MODEL_SUFFIXES = {"xgboost": "", "dnn": "_dnn", "mlp": "_mlp", "gnn": "_gnn"}

# Fixed process -> numeric id, same order as FILES in the master pipelines.
# Written as `process_id` so the files stay self-describing without a string
# branch (which uproot cannot write into a TTree, and TRExFitter cannot cut on).
PROCESS_IDS = {
    "signal_ggF": 0, "signal_VBF": 1, "Diboson": 2, "Zjets": 3, "Wjets": 4,
    "ttbar": 5, "tops": 6, "SingleH": 7, "Vgamma": 8, "VVV": 9,
}

PNN_COMBINED_CSV = BASE / "pnn_channel_combined" / "combined" / "oof_scores_combined.csv"

# Row-identity check before the positional concat. `run` is EXCLUDED on
# purpose: the per-run OOF csvs are inconsistent about carrying it (XGBoost
# 1l2tau/run2 has it, the DNN/GNN ones do not), while every combined-track csv
# does. Including it would break the comparison for no gain.
KEY_COLS = ["process", "label", "fold", "eventNumber", "w_phys"]

# Fixed dtypes for the non-score branches (scores are all float64).
BOOKKEEPING_DTYPES = {
    "eventNumber": np.int64,
    "fold": np.int32,
    "label": np.int32,
    "process_id": np.int32,
    "w_phys": np.float64,
}

print(f"uproot {uproot.__version__} | tree name {TREE_NAME!r} | "
      f"{len(CHANNELS)} channels x {len(RUNS)} runs x {len(PROCESS_IDS)} processes")
````

````python
_pnn_cache = None


def _read_csv(path):
    """Read an OOF csv, or return None if that notebook has not been run yet."""
    return pd.read_csv(path) if Path(path).exists() else None


def load_per_run_scores(channel, run, suffix):
    """Per-run-track OOF: one csv per (channel, run, model)."""
    return _read_csv(BASE / channel / run / f"oof_scores_{run}{suffix}.csv")


def load_combined_slice(channel, run, suffix):
    """
    Combined-track OOF: ONE csv per (channel, model) covering Run 2 and Run 3
    at once, carrying its own `run` column - so slice it rather than looking it
    up by filename. Verified row-for-row aligned with the per-run csv for all
    four (channel, run) combinations.
    """
    df = _read_csv(BASE / channel / "combined" / f"oof_scores_combined{suffix}.csv")
    if df is None:
        return None
    sub = df.loc[df["run"] == int(run.replace("run", ""))]
    return sub.reset_index(drop=True) if len(sub) else None


def load_pnn_slice(channel, run):
    """
    PNN writes ONE channel-combined OOF file covering both channels and both
    runs at once (columns include `channel`/`run`), so slice the cached csv.
    """
    global _pnn_cache
    if not PNN_COMBINED_CSV.exists():
        return None
    if _pnn_cache is None:
        _pnn_cache = pd.read_csv(PNN_COMBINED_CSV)
    sub = _pnn_cache.loc[
        (_pnn_cache["channel"] == channel)
        & (_pnn_cache["run"] == int(run.replace("run", "")))
    ]
    return sub.reset_index(drop=True) if len(sub) else None


def assemble_scores(channel, run):
    """
    Build ONE DataFrame for a (channel, run): the KEY_COLS bookkeeping columns
    taken once from the first source found, plus one `score_*` column per
    (model, track) that exists on disk.

    Scores are concatenated POSITIONALLY - NOT merged on `eventNumber`, which
    is not a unique key. The asserts below fail loudly instead of producing a
    corrupted export if that alignment is ever broken (e.g. two notebooks run
    with different preselections).

    Returns (merged_df or None, found, missing).
    """
    ref = merged = None
    found, missing = [], []

    def _add(col, df, name):
        nonlocal ref, merged
        if df is None or "score" not in df.columns:
            missing.append(name)
            return
        df = df.reset_index(drop=True)
        if merged is None:
            ref = df
            merged = df[KEY_COLS].copy()
        else:
            assert len(df) == len(ref), (
                f"{name}: {len(df)} rows vs {len(ref)} for the reference "
                f"({found[0]}) at {channel}/{run} - OOF files are no longer "
                f"row-aligned, cannot concatenate positionally.")
            assert ref[KEY_COLS].equals(df[KEY_COLS]), (
                f"{name}: KEY_COLS differ row-for-row from the reference "
                f"({found[0]}) at {channel}/{run} - the OOF files were not "
                f"written from the same source DataFrame, cannot concatenate "
                f"positionally.")
        merged[col] = df["score"].to_numpy(np.float64)
        found.append(name)

    for model, suffix in MODEL_SUFFIXES.items():
        _add(f"score_{model}", load_per_run_scores(channel, run, suffix), model)
    for model, suffix in MODEL_SUFFIXES.items():
        _add(f"score_{model}_comb", load_combined_slice(channel, run, suffix), f"{model}(comb)")
    _add("score_pnn", load_pnn_slice(channel, run), "pnn")

    return merged, found, missing
````

````python
def write_process_trees(merged, channel, run, tree_name=TREE_NAME):
    """
    Split `merged` by process and write ONE ROOT file per process, each a flat
    `TTree` (uproot.recreate + mktree + extend).

    `mktree` is what makes this a TTree - assigning `f[name] = dict` instead
    produces a `ROOT::RNTuple` in uproot >= 5.7, which TRExFitter cannot read.
    It also refuses string dtypes, which is why `process` becomes the numeric
    `process_id` and the process itself is encoded by the filename.

    Returns a list of per-process record dicts.
    """
    score_cols = [c for c in merged.columns if c.startswith("score_")]
    out_dir = BASE / channel / run
    out_dir.mkdir(parents=True, exist_ok=True)
    written = []

    for process, pid in PROCESS_IDS.items():
        mask = merged["process"] == process
        if not mask.any():
            warnings.warn(f"[{channel}/{run}] no rows for process {process!r} - no file written.")
            continue
        sub = merged.loc[mask]

        branches = {
            "eventNumber": sub["eventNumber"].to_numpy(np.int64),
            "fold":        sub["fold"].to_numpy(np.int32),
            "label":       sub["label"].to_numpy(np.int32),
            "process_id":  np.full(len(sub), pid, np.int32),
            "w_phys":      sub["w_phys"].to_numpy(np.float64),
        }
        for col in score_cols:
            branches[col] = sub[col].to_numpy(np.float64)

        out_path = out_dir / f"scores_{process}.root"
        with uproot.recreate(out_path) as f:
            f.mktree(tree_name, {k: v.dtype for k, v in branches.items()})
            f[tree_name].extend(branches)

        written.append({
            "channel": channel, "run": run, "process": process, "process_id": pid,
            "n_events": len(sub), "yield": float(sub["w_phys"].sum()),
            "n_score_branches": len(score_cols), "path": str(out_path),
        })

    return written
````

````python
all_written, region_rows = [], []

for channel in CHANNELS:
    for run in RUNS:
        merged, found, missing = assemble_scores(channel, run)

        if merged is None:
            print(f"[{channel}/{run}] NO model scores found at all (missing={missing}) - skipped.")
            region_rows.append({"channel": channel, "run": run, "n_events": 0, "n_files": 0,
                                "models_found": "-", "models_missing": ", ".join(missing)})
            continue

        written = write_process_trees(merged, channel, run)
        all_written.extend(written)

        print(f"[{channel}/{run}] {len(merged):>9,} rows -> {len(written):2d} files | "
              f"found: {found}")
        if missing:
            warnings.warn(f"[{channel}/{run}] missing OOF scores for: {missing} - those "
                          f"score_* branches are ABSENT from this export (not sentinel-filled) "
                          f"until the corresponding notebook is executed.")

        region_rows.append({"channel": channel, "run": run, "n_events": len(merged),
                            "n_files": len(written), "models_found": ", ".join(found),
                            "models_missing": ", ".join(missing) or "-"})

region_summary = pd.DataFrame(region_rows)
file_summary = pd.DataFrame(all_written)
print("\n=== per-region summary ===")
print(region_summary.to_string(index=False))
````

````python
# ---- Verification: everything TRExFitter needs, checked against the source csvs.
# Compiling/writing is not enough - re-open each file and assert (a) it is a
# real TTree, (b) no branch is a string, (c) row counts and yields round-trip.
n_ok = 0
problems = []

for rec in all_written:
    path = Path(rec["path"])
    with uproot.open(path) as f:
        classnames = list(f.classnames().values())
        if classnames != ["TTree"]:
            problems.append(f"{path}: classname {classnames} != ['TTree'] "
                            f"- TRExFitter cannot read this")
            continue
        tree = f[TREE_NAME]

        str_branches = [k for k, v in tree.typenames().items()
                        if "string" in str(v) or "char" in str(v)]
        if str_branches:
            problems.append(f"{path}: string branch(es) {str_branches} - "
                            f"TRExFitter Selection cannot cut on these")

        if tree.num_entries != rec["n_events"]:
            problems.append(f"{path}: {tree.num_entries} entries != {rec['n_events']} expected")

        w = tree["w_phys"].array(library="np")
        if not np.isclose(w.sum(), rec["yield"], rtol=1e-12, atol=1e-12):
            problems.append(f"{path}: yield {w.sum():.10g} != {rec['yield']:.10g}")

        n_ok += 1

print(f"Verified {n_ok} / {len(all_written)} files")
if problems:
    print("\nPROBLEMS:")
    for p in problems:
        print("  -", p)
    raise AssertionError(f"{len(problems)} verification failure(s) - see above")
print("All files are TTrees, string-free, and round-trip their row counts and yields.")

if all_written:
    example = Path(all_written[0]["path"])
    with uproot.open(example) as f:
        print(f"\nBranches in {example}:")
        for k, v in f[TREE_NAME].typenames().items():
            print(f"   {k:24} {v}")
    print("\n=== per-file summary ===")
    print(file_summary.drop(columns=["path"]).to_string(index=False))
````
